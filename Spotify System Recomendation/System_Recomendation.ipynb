```python
# Menghapus kolom 'Unnamed: 0' 
# Fungsi .drop() digunakan untuk membuang data.
# Parameter axis=1 memberi tahu pandas bahwa yang ingin kita buang adalah sebuah KOLOM (jika axis=0 berarti BARIS).

dataset = dataset.drop(['Unnamed: 0'],axis=1)

dataset.head()
```

```python
#Menghapus missing values

dataset.isnull().sum()
dataset = dataset.dropna()

print(dataset.isnull().sum())
```

```python
#Hapus duplikat
dataset.duplicated().sum()
dataset = dataset.sort_values('popularity', ascending=False)
dataset = dataset.drop_duplicates(subset=['track_name','artists'], keep='first')
dataset = dataset.reset_index(drop=True)
print(f"Jumlah data tersisa setelah duplikat dihapus: {dataset.shape[0]}")
```

```python
# Menggabungkan kolom 'track_genre' dan 'artists' menjadi satu kolom baru bernama 'tags'.
# Setiap kata akan dipisahkan dengan spasi.
dataset['tags'] = dataset['track_genre'] + " " + dataset['artists']

# Tampilkan beberapa baris untuk melihat hasilnya
dataset[['track_name', 'artists', 'track_genre', 'tags']].head()
```

```python
# Inisialisasi TfidfVectorizer
# stop_words='english' membuang kata-kata penghubung umum dalam bahasa Inggris yang tidak penting (seperti "and", "the")
tfidf = TfidfVectorizer(stop_words='english')

# Memproses (fit) dan mengubah (transform) teks di kolom 'tags' menjadi matriks
tfidf_matrix = tfidf.fit_transform(dataset['tags'])

# Melihat ukuran matriks yang dihasilkan
print(f"Bentuk matriks TF-IDF: {tfidf_matrix.shape}")
# Outputnya nanti (Jumlah Lagu, Jumlah Kata Unik dari seluruh Genre dan Artis)
```

```python
# Membuat "buku panduan" (Series) untuk memudahkan kita mencari index baris berdasarkan Judul Lagu
indices = pd.Series(dataset.index, index=dataset['track_name']).drop_duplicates()

# Membuat fungsi utama untuk sistem rekomendasi
def get_recommendations(title, tfidf_matrix=tfidf_matrix, df=dataset, indices=indices):
    # 1. Cek apakah lagu ada di dataset
    if title not in indices:
        return "Lagu tidak ditemukan. Pastikan huruf besar/kecilnya sama persis."
        
    # 2. Ambil index dari lagu yang dicari
    idx = indices[title]
    
    # Berjaga-jaga jika ada lagu duplikat dengan judul sama, kita ambil index yang pertama
    if type(idx) == pd.Series:
        idx = idx.iloc[0]
        
    # 3. Hitung skor kemiripan HANYA untuk lagu ini (menghindari memory error/RAM penuh)
    # linear_kernel (cosine similarity) akan membandingkan lagu ini vs seluruh lagu lainnya
    sim_scores = linear_kernel(tfidf_matrix[idx], tfidf_matrix).flatten()
    
    # 4. Urutkan urutan index lagu dari skor yang paling mirip (tertinggi) ke terendah
    top_indices = sim_scores.argsort()[::-1]
    
    # 5. Ambil 10 lagu teratas yang paling mirip. 
    # (Index ke-0 selalu lagu itu sendiri, jadi kita ambil urutan ke-1 sampai ke-10)
    top_10_indices = top_indices[1:11]
    
    # 6. Kembalikan data berupa Judul, Artis, dan Genre dari lagu-lagu rekomendasi
    return df[['track_name', 'artists', 'track_genre']].iloc[top_10_indices]
```

```python
# Tentukan judul lagu yang ingin dicari
judul_lagu = 'What You Need'

# Cetak teks cantiknya
print(f"\n🎵 Rekomendasi lagu mirip '{judul_lagu}':")

# Panggil fungsi yang tadi sudah kita buat (ditambah .head() agar rapi)
get_recommendations(judul_lagu).head()
```
